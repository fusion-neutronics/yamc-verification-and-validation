# D1S Decay-Photon Transport - ENDF/B-VIII.1

This notebook compares YAMC and OpenMC D1S (Direct 1-Step) decay-photon transport results using a sphere geometry with the ENDF/B-VIII.1 nuclear data library.

## Model Description

- **Geometry**: 35 cm radius sphere with vacuum boundary
- **Source**: Point source at origin, 14.06 MeV monoenergetic neutrons, isotropic
- **Material**: Single nuclide at 1.0 g/cm³, temperature 294 K
- **Transport**: Coupled neutron-photon with decay photons (`use_decay_photons=True`)
- **Statistics**: 500,000 particles × 10 batches = 5,000,000 total histories
- **Timesteps**: 3 × 1 hour irradiation + 2 × 15 day cooling
- **Tallies**: TCF-corrected photon flux spectra (VITAMIN-J-42) at each timestep

```python
# Setup: locate the yamc_verify package (repo root) and load this task.
import sys
from pathlib import Path as _Path
_repo = _Path.cwd().resolve()
while _repo != _repo.parent and not (_repo / "yamc_verify").exists():
    _repo = _repo.parent
if str(_repo) not in sys.path:
    sys.path.insert(0, str(_repo))

from yamc_verify import backends, get_task
from yamc_verify.families import decay_photon

TASK = get_task("decay-photon/sphere/endf-b8.1")
BACKENDS = backends.resolve()
print(f"Backends: {BACKENDS.tokens}")
```

```python
# Run all backends for every discovered nuclide (OpenMC reference cached on disk).
all_results, first_yamc_model = decay_photon.run_all(TASK, BACKENDS)
```
