# Broomstick Model - FENDL-3.2c

This notebook compares YAMC and OpenMC simulation results for a broomstick (cylinder) geometry using the FENDL-3.2c nuclear data library for neutron transport.

## Model Description

- **Geometry**: Cylinder with radius 1 cm, height 100 cm (z = -50 to +50), vacuum boundaries
- **Source**: Point source at origin, 14.06 MeV monoenergetic neutrons, isotropic
- **Material**: Single nuclide at 1.0 g/cm³, temperature 294 K
- **Statistics**: 500,000 particles × 10 batches = 5,000,000 total histories
- **Tallies**: Flux, heating, and all available reaction MTs (energy-integrated and energy-binned)

```python
# Setup: locate the yamc_verify package (repo root) and load this task.
import sys
from pathlib import Path as _Path
_repo = _Path.cwd().resolve()
while _repo != _repo.parent and not (_repo / "yamc_verify").exists():
    _repo = _repo.parent
if str(_repo) not in sys.path:
    sys.path.insert(0, str(_repo))

from yamc_verify import backends, get_task
from yamc_verify.families import neutron

TASK = get_task("neutron/broomstick/fendl-3.2c")
BACKENDS = backends.resolve()
print(f"Backends: {BACKENDS.tokens}")
```

```python
# Run all backends for every discovered nuclide (results cached on disk).
all_results, first_yamc_model = neutron.run_all(TASK, BACKENDS)
```

## Geometry

```python
# Geometry preview from the first successful YAMC model.
neutron.display_geometry(first_yamc_model, TASK)
```

## Results by Nuclide

```python
# Per-nuclide flux-spectra comparison, agreement metrics, and tally tables.
neutron.display_results(TASK, all_results)
```
